```python
import json
import matplotlib.pyplot as plt

# Step 1: Load JSON lines from file
with open('/scratch/radjoe/Projects/Experiments/SAMEXP/logs/decoder_only_adapter/logs/train_stats.json', 'r') as f:
    data = [json.loads(line) for line in f]

# with open('/scratch/radjoe/Projects/Experiments/SAMEXP/logs/decoder_only/logs/val_stats.json', 'r') as f:
#     val_data = [json.loads(line) for line in f]

# Step 2: Extract metrics over epochs
epochs = [entry["Trainer/epoch"] for entry in data]
loss_all = [entry["Losses/train_all_loss"] for entry in data]
loss_dice = [entry["Losses/train_all_loss_dice"] for entry in data]
loss_iou = [entry["Losses/train_all_loss_iou"] for entry in data]
loss_class = [entry["Losses/train_all_loss_class"] for entry in data]
loss_mask = [entry["Losses/train_all_loss_mask"] for entry in data]

# val_all = [entry["Losses/val_all_loss"] for entry in val_data]
# Step 3: Plotting
def sliding_moving_average(lst, window_size=3):
    return [
        sum(lst[i:i+window_size]) / window_size
        for i in range(len(lst) - window_size + 1)
    ]
plt.figure(figsize=(12, 7))

plt.plot(epochs[6:], sliding_moving_average(loss_all, window_size=3)[4:], label="Total Loss", linewidth=2)
# plt.plot(epochs, sliding_moving_average(loss_dice, window_size=3), label="Dice Loss")
# plt.plot(epochs, sliding_moving_average(loss_iou, window_size=3), label="IoU Loss")
# plt.plot(epochs, sliding_moving_average(loss_class, window_size=3), label="Class Loss")
# plt.plot(epochs, sliding_moving_average(loss_mask, window_size=3), label="Mask Loss")
# plt.plot(epochs, val_all, label="Val Loss")


plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss over Epochs")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
