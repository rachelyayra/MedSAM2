## The foreground bins

```python
def analyze_labels(npy_files):
    # Check for presence of classes in each file
    # print("Numpy files shape:", npy_files[1:].sum(axis=(3)).shape)
    npy_files = npy_files[1:,:,:, 50 : 120].sum(axis=(3))
    label_presence = npy_files.any(axis=(1,2)).astype(int)
    # print("Label presence per class:\n", label_presence)

    # Calculate foreground proportion per class
    per_class_foreground = []
    for i in range(npy_files.shape[0]):
        foreground = npy_files[i].sum()
        # print(f"Foreground {foreground}")
        total = 192 * 224 * 70  # Assuming fixed dimensions
        proportion = foreground / total
        per_class_foreground.append(proportion)
    per_class_foreground = np.array(per_class_foreground)
    # print("Foreground proportion per class:\n", per_class_foreground)
    label_info = np.concatenate((per_class_foreground, label_presence))
    return label_info
```

```python
test_npy = np.load('/scratch_net/ken/radjoe/BraTS_SSA/Validation/labels_UNN/BraTS-SSA-00015-000-lbl.npy')
```

```python
analyze_labels(test_npy)
```

```python
df = pd.DataFrame(columns=['Class_1_FG_Proportion', 'Class_2_FG_Proportion', 'Class_3_FG_Proportion', 'Class_1_Presence', 'Class_2_Presence', 'Class_3_Presence'])
for i in os.listdir(data_set_path):
    if i.endswith('.npy'):
        npy_file = np.load(os.path.join(data_set_path, i))
        label_info = analyze_labels(npy_file)
        # label_info = np.concatenate((np.array([i]), label_info))
        df.loc[i] = label_info
```

```python
def save_split(folder_path, label_path, save_path, train_files, cross_val):
    
    img_files = []

    for lbl in os.listdir(folder_path):
        for img in cross_val:
            print(img[:-8], lbl[:-8])
            if img[:-8] == lbl[:-8]:
                img_files.append(lbl)

    trainimg_files = []
    for lbl in os.listdir(folder_path):
        for img in train_files:
            print(img[:-8], lbl[:-8])
            if img[:-8] == lbl[:-8]:
                trainimg_files.append(lbl)

    val_savepath = f'{save_path}/Test/images_UNN'
    label_savepath = f'{save_path}/Test/labels_UNN'
    train_savepath = f'{save_path}/Validation/images_UNN'
    trainlabel_savepath = f'{save_path}/Validation/labels_UNN'
    os.makedirs(val_savepath, exist_ok=True)
    os.makedirs(label_savepath, exist_ok=True)
    os.makedirs(train_savepath, exist_ok=True)
    os.makedirs(trainlabel_savepath, exist_ok=True)

    for file in trainimg_files:
        shutil.copy(os.path.join(folder_path, file), val_savepath)

    for file in train_files:
        shutil.copy(os.path.join(label_path, file), label_savepath)

    for file in img_files:
        shutil.copy(os.path.join(folder_path, file), train_savepath)

    for file in cross_val:
        shutil.copy(os.path.join(label_path, file), trainlabel_savepath)
```
