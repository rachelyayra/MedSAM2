```python
path_to_img = '/scratch/radjoe/Preprocess_BraTS2023_SSA_Training/images_UNN/BraTS-SSA-00092-000-stk.npy'
```

```python
def renormalize(arr):
    foreground = arr[arr != 0]
    print(np.unique(foreground))
    print(foreground.shape)
    min_val, max_val = np.min(foreground), np.max(foreground)

    scaled = (arr - min_val) / (max_val - min_val + 1e-8)
    scaled[arr == 0] = 0 

    return scaled
```

```python
images = np.load(path_to_img)

images = images[1:, :, :, :]
for i in range(images.shape[0]):
        images[i] = renormalize(images[i])
images = torch.from_numpy(images)
```
